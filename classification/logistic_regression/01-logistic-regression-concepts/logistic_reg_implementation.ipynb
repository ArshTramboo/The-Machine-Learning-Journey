```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline 
```

```python
from sklearn.datasets import make_classification
```

```python
x,y=make_classification(n_samples=1000,n_features=10,n_classes=2,random_state=42)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.30,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
logistic=LogisticRegression()
```

```python
logistic.fit(x_train,y_train)
```

```python
y_pred=logistic.predict(x_test)
y_pred
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
```

```python
score=accuracy_score(y_test,y_pred)
from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_predictions(y_test, y_pred)
print(score)

print(classification_report(y_test,y_pred))
```

```python
model = LogisticRegression(max_iter=1000)

param_grid = [
    {'penalty': ['l2'], 'solver': ['newton-cg', 'lbfgs', 'sag', 'liblinear'], 'C': [100, 10, 1.0, 0.1, 0.01]},
    {'penalty': ['l1'], 'solver': ['liblinear', 'saga'], 'C': [100, 10, 1.0, 0.1, 0.01]},
    {'penalty': ['elasticnet'], 'solver': ['saga'], 'l1_ratio': [0.5], 'C': [100, 10, 1.0, 0.1, 0.01]},
]

from sklearn.model_selection import StratifiedKFold, GridSearchCV
cv = StratifiedKFold()

grid = GridSearchCV(estimator=model, param_grid=param_grid, scoring='accuracy', cv=cv, n_jobs=-1)
```

```python
grid
```

```python
grid.fit(x_train,y_train)
```

```python


print("Best Params:", grid.best_params_)
print("Best Score:", grid.best_score_)
```

```python
y_pred=grid.predict(x_test)
```

```python
score=accuracy_score(y_test,y_pred)
from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_predictions(y_test, y_pred)
print(score)

print(classification_report(y_test,y_pred))
```
